### 🔮 Think Before You Run (Prediction)
* **Question:** Which outcomes do you expect? Will the two qubits still produce the same outcomes? Will $00$ and $11$ dominate? Could $01$ and $10$ become more frequent?
* **Prediction:** If we add a Pauli-$X$ (NOT) gate to one of the qubits right after preparing the standard Bell state, it will flip the state of that qubit. Therefore, matching outcomes ($00$ and $11$) will no longer dominate. Instead, outcomes $01$ and $10$ will become more frequent because whenever qubit 0 measures $0$, qubit 1 is flipped to $1$ (and vice versa).

```python
from qiskit import QuantumCircuit
from qiskit.visualization import plot_histogram
from qiskit.providers.basic_provider import BasicProvider
from IPython.display import display

print("=== PART 1: ORIGINAL CIRCUIT ===")

# 1. Create a two-qubit quantum circuit with 2 classical bits
qc_original = QuantumCircuit(2, 2)

# 2. Start from the Bell-state circuit demonstrated during the workshop
qc_original.h(0)
qc_original.cx(0, 1)

# 4. Measure both qubits
qc_original.measure([0, 1], [0, 1])

# Display Original Circuit diagram (Submission Requirement)
print("\n[Original Circuit Diagram]:")
display(qc_original.draw('mpl'))

# 5. Execute the circuit for 100 shots using BasicProvider
provider = BasicProvider()
backend = provider.get_backend("basic_simulator")
job_orig = backend.run(qc_original, shots=100)
result_orig = job_orig.result()

# 6. Observe the measurement results
counts_orig = result_orig.get_counts(qc_original)
print("\n[Original Circuit Measurement Results - 100 shots]:", counts_orig)

# 7. Display the results using a histogram
print("\n[Original Circuit Histogram]:")
display(plot_histogram(counts_orig))
```

```python
print("\n=== PART 2: MODIFIED CIRCUIT ===")

# 1 & 2. Create a two-qubit quantum circuit starting from the Bell-state circuit
qc_modified = QuantumCircuit(2, 2)
qc_modified.h(0)
qc_modified.cx(0, 1)

# 3. Add or modify one quantum gate (Adding X gate on qubit 1)
# Design Constraint Check: Uses 2 qubits, measures both, exactly 3 gates before measurement (< 4 limit)
qc_modified.x(1)

# 4. Measure both qubits
qc_modified.measure([0, 1], [0, 1])

# Display Modified Circuit diagram (Submission Requirement)
print("\n[Modified Circuit Diagram]:")
display(qc_modified.draw('mpl'))

# 5. Execute the circuit for 100 shots
job_mod = backend.run(qc_modified, shots=100)
result_mod = job_mod.result()

# 6. Observe the measurement results
counts_mod = result_mod.get_counts(qc_modified)
print("\n[Modified Circuit Measurement Results - 100 shots]:", counts_mod)

# 7. Display the results using a histogram
print("\n[Modified Circuit Histogram]:")
display(plot_histogram(counts_mod))
```

* **8. Comparison with the Original Circuit:**
  * **Original Circuit:** The standard Bell-state circuit produced measurements heavily dominated by matching outcomes ($00$ and $11$), showing strong positive correlation ($\frac{|00\rangle + |11\rangle}{\sqrt{2}}$).
  * **Modified Circuit:** The modified circuit produced measurements dominated by orthogonal/flipped outcomes ($01$ and $10$).

* **9. Explanation of the Observed Change:**
  By adding a single Pauli-$X$ (NOT) gate to qubit 1 immediately after the entangling $CNOT$ gate, its logical measurement outcome is inverted. This successfully altered the correlation relationship between the two measurement outcomes.
