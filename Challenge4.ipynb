## Think Before You Run: Predictions
* **Part A Predictions:** We expect outcomes `00` and `11` to dominate (roughly 50/50 split) because the circuit builds an entangled Bell state ($\frac{\vert{}00\rangle + \vert{}11\rangle}{\sqrt{2}}$).
* **Part B Predictions:** We expect outcomes `01` and `10` to dominate because adding a Pauli-X gate flips the second qubit's outcome.
* **Will all four possible outcomes appear?** In an ideal noiseless simulator with 100 shots, only the two valid states for each respective circuit will appear. Minor noise or imperfections might show tiny values on real hardware, but simulator results will show just 2 dominant outcomes.

```python
# Ensure matplotlib renders colorful graphical circuit diagrams inside the notebook
%matplotlib inline

from qiskit import QuantumCircuit
from qiskit.providers.basic_provider import BasicProvider
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Initialize backend
provider = BasicProvider()
backend = provider.get_backend("basic_simulator")

# ==========================================
# PART A: Correlated Outcomes
# ==========================================
qc_a = QuantumCircuit(2, 2)
qc_a.h(0)       # Hadamard gate on qubit 0
qc_a.cx(0, 1)   # CNOT gate with control 0, target 1
qc_a.measure([0, 1], [0, 1])

print("Circuit A Diagram:")
# Display full-color graphical circuit diagram
display(qc_a.draw('mpl'))

# Run simulation (100 shots)
job_a = backend.run(qc_a, shots=100)
counts_a = job_a.result().get_counts(qc_a)
print("\nPart A Counts (100 shots):", counts_a)
```

```python
# ==========================================
# PART B: Opposite Outcomes
# ==========================================
qc_b = QuantumCircuit(2, 2)
qc_b.h(0)       # Hadamard gate on qubit 0
qc_b.cx(0, 1)   # CNOT gate with control 0, target 1
qc_b.x(1)       # Pauli-X gate on qubit 1 to flip outcome
qc_b.measure([0, 1], [0, 1])

print("Circuit B Diagram:")
# Display full-color graphical circuit diagram
display(qc_b.draw('mpl'))

# Run simulation (100 shots)
job_b = backend.run(qc_b, shots=100)
counts_b = job_b.result().get_counts(qc_b)
print("\nPart B Counts (100 shots):", counts_b)
```

```python
# ==========================================
# PART C: Histograms Comparison
# ==========================================
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
plot_histogram(counts_a, ax=ax[0], title="Part A: Correlated (00 & 11)")
plot_histogram(counts_b, ax=ax[1], title="Part B: Opposite (01 & 10)")
plt.tight_layout()
plt.show()
```

## Part C: Comparison and Explanations

1. **Which outcomes are dominant in Part A?**
   * `00` and `11` are dominant (each appearing roughly 50 times out of 100 shots).

2. **Which outcomes are dominant in Part B?**
   * `01` and `10` are dominant (each appearing roughly 50 times out of 100 shots).

3. **What changed between the two circuits?**
   * A **Pauli-X (NOT) gate** was added to qubit 1 right before measurement in Part B. Both circuits successfully obeyed the constraint of using exactly 2 qubits, measuring both qubits, and using no more than 4 quantum gates.

4. **Why did the measurement relationship change?**
   * The Hadamard and CNOT gates create a maximally entangled Bell state ($\frac{\vert{}00\rangle + \vert{}11\rangle}{\sqrt{2}}$), which forces both qubits to measure to the exact same value. Inserting the $X$ gate acts as a bit-flip on the second qubit ($X\vert{}0\rangle = \vert{}1\rangle$ and $X\vert{}1\rangle = \vert{}0\rangle$), inverting its bit value and switching the measurement correlation from matching (`00`/`11`) to opposite (`01`/`10`).
