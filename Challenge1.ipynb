## 1. Think Before You Run (Predictions)
* **What will happen after applying one X gate?** 
  * The $X$ gate acts as a quantum NOT gate, flipping the initial $\vert{}0\rangle$ state to $\vert{}1\rangle$. Therefore, the measurement result will always be **`1`** (100% of the time).
* **What will happen after applying two X gates?** 
  * Applying two $X$ gates in succession ($\text{NOT} \times \text{NOT}$) returns the qubit back to its original state $\vert{}0\rangle$. Therefore, the measurement result will always be **`0`** (100% of the time).
* **What do you expect after applying an H gate? Will Circuit C always produce the same measurement result?** 
  * The Hadamard ($H$) gate creates an equal superposition state: $\frac{1}{\sqrt{2}}(\vert{}0\rangle + \vert{}1\rangle)$. **No**, it will not always produce the same measurement result; it will yield a probabilistic distribution split roughly 50/50 between **`0`** and **`1`** over 100 shots.

```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler as Sampler
from qiskit.visualization import plot_histogram

# --- Design Constraints Check ---
# 1. Exactly 1 qubit used per circuit
# 2. Initial state is |0> by default in QuantumCircuit(1, 1)
# 3. Measurement is included

# --- Circuit A: Apply an X gate and measure ---
qc_a = QuantumCircuit(1, 1)
qc_a.x(0)          # Apply X gate
qc_a.measure(0, 0) # Measure qubit 0 into classical bit 0

# --- Circuit B: Apply two X gates and measure ---
qc_b = QuantumCircuit(1, 1)
qc_b.x(0)          # First X gate
qc_b.x(0)          # Second X gate
qc_b.measure(0, 0)

# --- Circuit C: Apply an H gate and measure ---
qc_c = QuantumCircuit(1, 1)
qc_c.h(0)          # Apply Hadamard gate
qc_c.measure(0, 0)

# Display the three quantum circuits
print("=== Circuit A (One X Gate) ===")
display(qc_a.draw('mpl'))

print("=== Circuit B (Two X Gates) ===")
display(qc_b.draw('mpl'))

print("=== Circuit C (Hadamard Gate) ===")
display(qc_c.draw('mpl'))

# Execute each circuit for 100 shots using Sampler
sampler = Sampler()
job = sampler.run([qc_a, qc_b, qc_c], shots=100)
result = job.result()

# Record measurement results for 100 shots
counts_a = result[0].data.c.get_counts()
counts_b = result[1].data.c.get_counts()
counts_c = result[2].data.c.get_counts()

print("\n=== Measurement Results (100 Shots) ===")
print("Circuit A (X Gate) Counts:", counts_a)
print("Circuit B (Two X Gates) Counts:", counts_b)
print("Circuit C (H Gate) Counts:", counts_c)

# Display the results using a histogram
display(plot_histogram([counts_a, counts_b, counts_c], 
                       legend=['Circuit A (X)', 'Circuit B (XX)', 'Circuit C (H)'],
                       title="Quantum Circuit Measurement Results (100 Shots)"))
```
